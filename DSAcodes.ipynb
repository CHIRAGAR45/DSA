1. Insert a new element in array

```python

arr = [10, 20, 30, 40, 50]

print("Original array:", arr)

pos = int(input("Enter position: "))
item = int(input("Enter element: "))

for i in range(len(arr) - 1, pos - 1, -1):
    arr[i] = arr[i - 1]

arr[pos - 1] = item

print("Array after insertion:", arr)
```

2. Delete an element from array

```python

arr = [10, 20, 30, 40, 50]

print("Original array:", arr)

pos = int(input("Enter position to delete: "))

for i in range(pos - 1, len(arr) - 1):
    arr[i] = arr[i + 1]

arr = arr[:-1]

print("Array after deletion:", arr)
```

3. Find product of two matrix

```python
a = [[1, 2], [3, 4]]
b = [[5, 6], [7, 8]]

c = [[0, 0], [0, 0]]

for i in range(2):
    for j in range(2):
        for k in range(2):
            c[i][j] = c[i][j] + a[i][k] * b[k][j]

print("Product of two matrices:")
for row in c:
    print(row)
```

4. Sum of diagonal and non diagonal elements in matrix

```python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]

diagonal = 0
non_diagonal = 0

for i in range(3):
    for j in range(3):
        if i == j:
            diagonal += a[i][j]
        else:
            non_diagonal += a[i][j]

print("Sum of diagonal elements:", diagonal)
print("Sum of non-diagonal elements:", non_diagonal)
```

6. Stack menu-

```python
n = int(input("Enter size of stack: "))
stack = [0] * n
top = -1

while True:
    print("\n1. Create")
    print("2. Push")
    print("3. Pop")
    print("4. Display")
    print("5. Underflow")
    print("6. Overflow")
    print("7. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        stack = [0] * n
        top = -1
        print("Stack created")

    elif choice == 2:
        if top == n - 1:
            print("Overflow")
        else:
            x = int(input("Enter element: "))
            top = top + 1
            stack[top] = x

    elif choice == 3:
        if top == -1:
            print("Underflow")
        else:
            print("Popped:", stack[top])
            top = top - 1

    elif choice == 4:
        if top == -1:
            print("Stack is empty")
        else:
            for i in range(top, -1, -1):
                print(stack[i])

    elif choice == 5:
        if top == -1:
            print("Underflow")
        else:
            print("No Underflow")

    elif choice == 6:
        if top == n - 1:
            print("Overflow")
        else:
            print("No Overflow")

    elif choice == 7:
        break
2
```

7. Queue menu-

```python
size = int(input("Enter size of queue: "))
queue = [0] * size
front = -1
rear = -1

while True:
    print("\n1. Create")
    print("2. Enqueue")
    print("3. Dequeue")
    print("4. Display")
    print("5. Underflow")
    print("6. Overflow")
    print("7. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        # Re-initialize the queue to its original size
        queue = [0] * size
        front = -1
        rear = -1
        print("Queue created")

    elif choice == 2:  # Enqueue operation
        if rear == size - 1:
            print("Overflow: Queue is full")
        else:
            element = int(input("Enter element to enqueue: "))
            if front == -1:  # If queue was empty, set front to 0
                front = 0
            rear = rear + 1
            queue[rear] = element
            print(f"Enqueued: {element}")

    elif choice == 3:  # Dequeue operation
        if front == -1 or front > rear:
            print("Underflow: Queue is empty")
        else:
            print("Dequeued:", queue[front])
            front = front + 1
            if front > rear: # Reset if queue becomes empty after dequeue
                front = -1
                rear = -1

    elif choice == 4:  # Display operation
        if front == -1 or front > rear:
            print("Queue is empty")
        else:
            print("Queue elements:")
            for i in range(front, rear + 1):
                print(queue[i])

    elif choice == 5:  # Check Underflow
        if front == -1 or front > rear:
            print("Underflow")
        else:
            print("No Underflow")

    elif choice == 6:  # Check Overflow
        if rear == size - 1:
            print("Overflow")
        else:
            print("No Overflow")

    elif choice == 7:  # Exit
        break
```
